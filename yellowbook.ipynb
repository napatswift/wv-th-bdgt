```python
import pandas as pd
from thbud.model import BudgetItem, BudgetType
from anytree import PreOrderIter
import re
```

```python
df = pd.read_csv('./(เล่มเหลือง)yellow68.csv', index_col=0)
```

```python
rows = df.to_dict('records')
```

```python
def nm(value):
    if isinstance(value, str):
        return value.replace('PLANE', 'PLAN')
    return value
```

```python
rows = [
    {k: nm(v) for k, v in r.items() if not pd.isna(v)}
    for r in rows
]
```

```python
root = BudgetItem.build_tree_by_rows(rows)
```

```python
for node in PreOrderIter(root):
    if re.match('\d\. ยุทธศาสตร์',node.name,):
        node.budget_type = BudgetType.STRATEGY
    if node.budget_type == BudgetType.BUDGET_PLAN:
        for ministry in node.children:
            ministry.budget_type = BudgetType.MINISTRY
            for budget_unit in ministry.children:
                budget_unit.budget_type = BudgetType.BUDGETARY_UNIT
    if node.budget_type == BudgetType.BUDGET_DETAIL:
        print(node)
    pass
```
